The subtask requires creating a Python function for audio feature extraction. To facilitate this, I will now define the function using `librosa` for loading audio and generating Mel-spectrograms, including parameters for customization, as per the instructions.



````python
import librosa
import librosa.display
import numpy as np

def extract_mel_spectrogram(audio_path, sr=22050, n_mels=128, fmax=8000, hop_length=512):
    """
    Extracts a Mel-spectrogram from an audio file.

    Args:
        audio_path (str): Path to the audio file.
        sr (int): Sampling rate to resample the audio to. Defaults to 22050.
        n_mels (int): Number of Mel bands to generate. Defaults to 128.
        fmax (int): Maximum frequency to consider for Mel-spectrogram. Defaults to 8000.
        hop_length (int): The hop length for Mel-spectrogram calculation. Defaults to 512.

    Returns:
        np.ndarray: A Mel-spectrogram (log-scaled).
    """
    try:
        # Load the audio file
        y, sr = librosa.load(audio_path, sr=sr)

        # Generate Mel-spectrogram
        mel_spectrogram = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels, fmax=fmax, hop_length=hop_length)

        # Convert to decibels (log scale)
        mel_spectrogram_db = librosa.power_to_db(mel_spectrogram, ref=np.max)

        return mel_spectrogram_db
    except Exception as e:
        print(f"Error processing {audio_path}: {e}")
        return None

print("Mel-spectrogram extraction function 'extract_mel_spectrogram' defined.")
````

## Implement Data Augmentation Techniques

### Subtask:
Implement data augmentation techniques like Mixup, SpecAugment, and background noise mixing into the data loading pipeline to enhance model robustness and generalization.


**Reasoning**:
To implement the requested data augmentation techniques, I will define a main function `apply_audio_augmentation` that orchestrates three helper functions: one for Mixup, one for SpecAugment, and one for background noise mixing. This modular approach allows for clear implementation of each technique and fulfills all requirements of the subtask.



````python
def _mixup_spectrogram(spec1, labels1, spec2, labels2, alpha=0.2):
    """
    Performs Mixup augmentation on two Mel-spectrograms and their labels.

    Args:
        spec1 (np.ndarray): First Mel-spectrogram.
        labels1 (np.ndarray): Labels for the first spectrogram (e.g., one-hot encoded).
        spec2 (np.ndarray): Second Mel-spectrogram.
        labels2 (np.ndarray): Labels for the second spectrogram.
        alpha (float): Alpha parameter for the Beta distribution used to generate lambda.

    Returns:
        tuple: Augmented Mel-spectrogram and mixed labels.
    """
    lam = np.random.beta(alpha, alpha)
    augmented_spec = lam * spec1 + (1 - lam) * spec2
    augmented_labels = lam * labels1 + (1 - lam) * labels2
    return augmented_spec, augmented_labels

def _spec_augment_spectrogram(mel_spectrogram, freq_mask_param=20, time_mask_param=30, n_freq_masks=1, n_time_masks=1):
    """
    Applies SpecAugment (frequency and time masking) to a Mel-spectrogram.

    Args:
        mel_spectrogram (np.ndarray): The input Mel-spectrogram.
        freq_mask_param (int): Maximum width of the frequency mask.
        time_mask_param (int): Maximum width of the time mask.
        n_freq_masks (int): Number of frequency masks to apply.
        n_time_masks (int): Number of time masks to apply.

    Returns:
        np.ndarray: Augmented Mel-spectrogram.
    """
    augmented_spec = mel_spectrogram.copy()
    num_mels, num_frames = augmented_spec.shape

    # Frequency masking
    for _ in range(n_freq_masks):
        if freq_mask_param > 0:
            f = np.random.randint(0, freq_mask_param)
            f0 = np.random.randint(0, num_mels - f) if num_mels - f > 0 else 0
            augmented_spec[f0:f0 + f, :] = 0

    # Time masking
    for _ in range(n_time_masks):
        if time_mask_param > 0:
            t = np.random.randint(0, time_mask_param)
            t0 = np.random.randint(0, num_frames - t) if num_frames - t > 0 else 0
            augmented_spec[:, t0:t0 + t] = 0
    return augmented_spec

def _add_noise_to_spectrogram(mel_spectrogram, noise_mel_spectrogram, noise_strength=0.1):
    """
    Adds a background noise Mel-spectrogram to the input Mel-spectrogram.

    Args:
        mel_spectrogram (np.ndarray): The input Mel-spectrogram.
        noise_mel_spectrogram (np.ndarray): The Mel-spectrogram of the noise to add.
        noise_strength (float): The scaling factor for the noise.

    Returns:
        np.ndarray: Augmented Mel-spectrogram with noise.
    """
    if noise_mel_spectrogram is None or noise_strength == 0:
        return mel_spectrogram

    spec_shape = mel_spectrogram.shape
    noise_shape = noise_mel_spectrogram.shape

    # Ensure noise_mel_spectrogram has compatible number of Mel bands
    if noise_shape[0] != spec_shape[0]:
        # Resize noise_mel_spectrogram to match the number of Mel bands
        # This is a simplification; a proper approach might involve resampling Mel frequencies.
        # For now, we crop or pad.
        if noise_shape[0] > spec_shape[0]:
            noise_mel_spectrogram = noise_mel_spectrogram[:spec_shape[0], :]
        else:
            # Pad with zeros along the frequency axis
            padding = np.zeros((spec_shape[0] - noise_shape[0], noise_shape[1]))
            noise_mel_spectrogram = np.vstack([noise_mel_spectrogram, padding])
        noise_shape = noise_mel_spectrogram.shape # Update shape after padding/cropping

    # Ensure noise_mel_spectrogram has compatible time frames (crop or pad)
    if noise_shape[1] < spec_shape[1]:
        # Pad noise if it's shorter in time
        padding = np.zeros((spec_shape[0], spec_shape[1] - noise_shape[1]))
        processed_noise_spec = np.hstack([noise_mel_spectrogram, padding])
    elif noise_shape[1] > spec_shape[1]:
        # Crop noise if it's longer in time
        processed_noise_spec = noise_mel_spectrogram[:, :spec_shape[1]]
    else:
        processed_noise_spec = noise_mel_spectrogram

    augmented_spec = mel_spectrogram + noise_strength * processed_noise_spec
    return augmented_spec

def apply_audio_augmentation(
    mel_spectrogram,
    labels, # Assumed to be numerical (e.g., one-hot encoded) for Mixup
    mixup_enabled=False,
    mixup_target_spec=None, # Mel-spectrogram of another sample for Mixup
    mixup_target_labels=None, # Labels of another sample for Mixup
    mixup_alpha=0.2,
    spec_augment_enabled=False,
    freq_mask_param=20,
    time_mask_param=30,
    n_freq_masks=1,
    n_time_masks=1,
    noise_mix_enabled=False,
    noise_mel_spectrogram=None, # Mel-spectrogram of background noise
    noise_strength=0.1
):
    """
    Applies various data augmentation techniques to a Mel-spectrogram.

    Args:
        mel_spectrogram (np.ndarray): The input Mel-spectrogram.
        labels (np.ndarray): Labels corresponding to the Mel-spectrogram.
        mixup_enabled (bool): Whether to apply Mixup.
        mixup_target_spec (np.ndarray, optional): Mel-spectrogram of another sample for Mixup.
        mixup_target_labels (np.ndarray, optional): Labels of another sample for Mixup.
        mixup_alpha (float): Alpha parameter for Mixup.
        spec_augment_enabled (bool): Whether to apply SpecAugment.
        freq_mask_param (int): Max frequency mask width for SpecAugment.
        time_mask_param (int): Max time mask width for SpecAugment.
        n_freq_masks (int): Number of frequency masks for SpecAugment.
        n_time_masks (int): Number of time masks for SpecAugment.
        noise_mix_enabled (bool): Whether to apply background noise mixing.
        noise_mel_spectrogram (np.ndarray, optional): Mel-spectrogram of background noise.
        noise_strength (float): Strength of background noise to mix.

    Returns:
        tuple: Augmented Mel-spectrogram and potentially augmented labels.
    """
    augmented_spec = mel_spectrogram
    augmented_labels = labels

    if mixup_enabled and mixup_target_spec is not None and mixup_target_labels is not None:
        augmented_spec, augmented_labels = _mixup_spectrogram(
            augmented_spec, augmented_labels, mixup_target_spec, mixup_target_labels, mixup_alpha
        )

    if spec_augment_enabled:
        augmented_spec = _spec_augment_spectrogram(
            augmented_spec, freq_mask_param, time_mask_param, n_freq_masks, n_time_masks
        )

    if noise_mix_enabled and noise_mel_spectrogram is not None:
        augmented_spec = _add_noise_to_spectrogram(
            augmented_spec, noise_mel_spectrogram, noise_strength
        )

    return augmented_spec, augmented_labels

print("Data augmentation functions (_mixup_spectrogram, _spec_augment_spectrogram, _add_noise_to_spectrogram, and apply_audio_augmentation) defined.")
````

````python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import numpy as np
import pandas as pd
import ast
from sklearn.model_selection import train_test_split
import os
import librosa

# --- 0. Re-import and Process df_train (to ensure consistency from previous steps) ---
# Ensure train.csv is unzipped and loaded correctly, and secondary_labels are processed.
zip_file_path = '/content/birdclef-2026.zip'
print(f"Unzipping train.csv from {zip_file_path}...")
!unzip -o {zip_file_path} train.csv -d /content/
df_train = pd.read_csv('/content/train.csv')

# Process secondary_labels_processed again to ensure it's correct
df_train['secondary_labels_processed'] = df_train['secondary_labels'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)
df_train['secondary_labels_processed'] = df_train['secondary_labels_processed'].apply(lambda x: None if len(x) == 0 else x)

# --- 1. Prepare Label Mapping and Class Weights ---
# Re-calculate all unique labels based on df_train
unique_primary_labels = df_train['primary_label'].unique()
exploded_secondary_labels = df_train['secondary_labels_processed'].explode()
unique_secondary_labels = exploded_secondary_labels.dropna().unique()

all_unique_labels = sorted(list(set(unique_primary_labels) | set(unique_secondary_labels)))
num_classes = len(all_unique_labels)

label_to_idx = {label: i for i, label in enumerate(all_unique_labels)}
idx_to_label = {i: label for i, label in enumerate(all_unique_labels)}

# Calculate class weights for BCEWithLogitsLoss
# Count occurrences of each label (primary and secondary) across all samples
label_counts = {label: 0 for label in all_unique_labels}
for _, row in df_train.iterrows():
    # Primary label
    if row['primary_label'] in label_counts:
        label_counts[row['primary_label']] += 1
    # Secondary labels
    if row['secondary_labels_processed'] is not None:
        for label in row['secondary_labels_processed']:
            if label in label_counts:
                label_counts[label] += 1

total_samples = len(df_train)
# pos_weight for BCEWithLogitsLoss is (number of negative samples) / (number of positive samples)
pos_weight = torch.zeros(num_classes)
for label, count in label_counts.items():
    idx = label_to_idx[label]
    if count > 0:
        pos_weight[idx] = (total_samples - count) / count
    else:
        pos_weight[idx] = 1.0 # Or some large value if a class has zero positive samples

print(f"Total number of unique labels (num_classes): {num_classes}")
print(f"Class weights (pos_weight) calculated. First 10 elements: {pos_weight[:10].tolist()}")

# --- Re-define extract_mel_spectrogram (from 4ac0dc11) and BirdCLEF_CNN (from 320c0dfc) ---
# Updated extract_mel_spectrogram with consistent SR, n_fft, hop_length, and normalization
def extract_mel_spectrogram(audio_data, sr=32000, n_mels=128, fmax=8000, n_fft=1024, hop_length=512):
    """
    Extracts a Mel-spectrogram from an audio waveform.

    Args:
        audio_data (np.ndarray): Audio time series (waveform).
        sr (int): Sampling rate of the audio. Defaults to 32000.
        n_mels (int): Number of Mel bands to generate. Defaults to 128.
        fmax (int): Maximum frequency to consider for Mel-spectrogram. Defaults to 8000.
        n_fft (int): FFT window size. Defaults to 1024.
        hop_length (int): The hop length for Mel-spectrogram calculation. Defaults to 512.

    Returns:
        np.ndarray: A Mel-spectrogram (log-scaled).
    """
    try:
        # Normalize audio (optional, but good practice)
        y = librosa.util.normalize(audio_data)

        # Generate Mel-spectrogram
        mel_spectrogram = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels, fmax=fmax, n_fft=n_fft, hop_length=hop_length)

        # Convert to decibels (log scale)
        mel_spectrogram_db = librosa.power_to_db(mel_spectrogram, ref=np.max)

        return mel_spectrogram_db
    except Exception as e:
        # print(f"Error processing audio data: {e}") # Commented to reduce verbose output
        return None

class BirdCLEF_CNN(nn.Module):
    def __init__(self, num_classes, n_mels=128, num_time_steps=313):
        super(BirdCLEF_CNN, self).__init__()
        self.conv1 = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=(3, 3), padding=(1, 1)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2, 2))
        )
        self.conv2 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=(3, 3), padding=(1, 1)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2, 2))
        )
        self.conv3 = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=(3, 3), padding=(1, 1)),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=(2, 2))
        )

        with torch.no_grad():
            dummy_input = torch.randn(1, 1, n_mels, num_time_steps)
            x = self.conv1(dummy_input)
            x = self.conv2(x)
            x = self.conv3(x)
            self._to_linear = x.flatten().shape[0]

        self.fc = nn.Linear(self._to_linear, num_classes)

    def forward(self, x):
        x = self.conv1(x)
        x = self.conv2(x)
        x = self.conv3(x)
        x = x.flatten(1)
        x = self.fc(x)
        return x

# --- 2. Define BirdCLEFDataset Class ---
class BirdCLEFDataset(Dataset):
    def __init__(self, dataframe, audio_dir, label_to_idx, num_classes, n_mels=128, num_time_steps=313, sr=32000, target_length_sec=5):
        self.dataframe = dataframe
        self.audio_dir = audio_dir
        self.label_to_idx = label_to_idx
        self.num_classes = num_classes
        self.n_mels = n_mels
        self.num_time_steps = num_time_steps
        self.sr = sr
        self.target_samples = int(target_length_sec * sr)

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):
        row = self.dataframe.iloc[idx]
        audio_file_path = os.path.join(self.audio_dir, row['filename'])

        try:
            # Load audio file
            y, _ = librosa.load(audio_file_path, sr=self.sr, mono=True)

            # Pad/Crop audio to consistent size (TARGET_SAMPLES)
            if len(y) < self.target_samples:
                # Zero-pad if shorter
                padding = self.target_samples - len(y)
                y = np.pad(y, (0, padding), mode='constant')
            elif len(y) > self.target_samples:
                # Random crop if longer
                start_sample = np.random.randint(0, len(y) - self.target_samples + 1)
                y = y[start_sample : start_sample + self.target_samples]

        except Exception as e:
            # Handle cases where audio loading fails by returning a dummy tensor and zero labels
            # In a real scenario, you might want to log this or filter out such samples during dataset creation
            print(f"Error loading or processing audio file {audio_file_path}: {e}") # Added print for debugging
            dummy_spec = torch.zeros(1, self.n_mels, self.num_time_steps)
            dummy_labels = torch.zeros(self.num_classes)
            return dummy_spec, dummy_labels

        # Extract Mel-spectrogram from the processed audio waveform
        mel_spec = extract_mel_spectrogram(y, sr=self.sr, n_mels=self.n_mels, fmax=8000, n_fft=1024, hop_length=512)

        if mel_spec is None:
            dummy_spec = torch.zeros(1, self.n_mels, self.num_time_steps)
            dummy_labels = torch.zeros(self.num_classes)
            return dummy_spec, dummy_labels

        # Pad/Crop Mel-spectrogram to consistent size (num_time_steps)
        # This handles cases where hop_length might cause slight variations in frames
        if mel_spec.shape[1] < self.num_time_steps:
            # Pad with zeros
            pad_width = self.num_time_steps - mel_spec.shape[1]
            mel_spec = np.pad(mel_spec, ((0, 0), (0, pad_width)), mode='constant')
        elif mel_spec.shape[1] > self.num_time_steps:
            # Crop
            mel_spec = mel_spec[:, :self.num_time_steps]

        # Create multi-hot encoded label vector
        labels = torch.zeros(self.num_classes, dtype=torch.float32)

        # Primary label
        if row['primary_label'] in self.label_to_idx:
            labels[self.label_to_idx[row['primary_label']]] = 1.0

        # Secondary labels
        if row['secondary_labels_processed'] is not None:
            for secondary_label in row['secondary_labels_processed']:
                if secondary_label in self.label_to_idx:
                    labels[self.label_to_idx[secondary_label]] = 1.0

        # Add channel dimension for CNN (1, n_mels, num_time_steps)
        mel_spec_tensor = torch.tensor(mel_spec, dtype=torch.float32).unsqueeze(0)

        return mel_spec_tensor, labels

print("BirdCLEFDataset class defined.")

# --- 3. Create DataLoaders ---
# Split data into training and validation sets
train_df, val_df = train_test_split(df_train, test_size=0.2, random_state=42)

# Audio directory (assuming train_audio is extracted to /content/train_audio/)
audio_root_dir = '/content/train_audio/'
# Check if audio_root_dir exists
if not os.path.exists(audio_root_dir):
    print(f"Warning: {audio_root_dir} not found. Attempting to unzip train_audio folder.")
    print(f"Unzipping train_audio from {zip_file_path}...")
    # Unzip only the train_audio directory contents from the downloaded birdclef-2026.zip
    # This assumes train_audio is directly inside the zip. Might take a long time.
    !unzip -o {zip_file_path} "train_audio/*" -d /content/


# Instantiate datasets
train_dataset = BirdCLEFDataset(
    dataframe=train_df,
    audio_dir=audio_root_dir,
    label_to_idx=label_to_idx,
    num_classes=num_classes,
    n_mels=n_mels,
    num_time_steps=num_time_steps,
    sr=32000, # Ensure consistent SR
    target_length_sec=5
)
val_dataset = BirdCLEFDataset(
    dataframe=val_df,
    audio_dir=audio_root_dir,
    label_to_idx=label_to_idx,
    num_classes=num_classes,
    n_mels=n_mels,
    num_time_steps=num_time_steps,
    sr=32000, # Ensure consistent SR
    target_length_sec=5
)

batch_size = 32
num_workers = 2 # Adjust based on your environment

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=num_workers)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=num_workers)

print(f"Created training DataLoader with {len(train_dataset)} samples and validation DataLoader with {len(val_dataset)} samples.")

# --- 4. Define Loss Function and Optimizer ---
# Instantiate the model
model = BirdCLEF_CNN(num_classes, n_mels, num_time_steps)

# Move pos_weight to the device if using GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight.to(device))
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

print(f"Model initialized and moved to {device}.")
print(f"Loss function (BCEWithLogitsLoss with pos_weight) and Optimizer (Adam) defined.")

# --- 5. Implement Training Loop ---
num_epochs = 5 # Reduced for demonstration purposes

model.to(device)

print("Starting training loop...")
for epoch in range(num_epochs):
    model.train() # Set model to training mode
    running_loss = 0.0
    for i, (inputs, labels) in enumerate(train_loader):
        inputs = inputs.to(device)
        labels = labels.to(device)

        optimizer.zero_grad() # Zero the parameter gradients

        outputs = model(inputs) # Forward pass
        loss = criterion(outputs, labels) # Calculate loss
        loss.backward() # Backward pass
        optimizer.step() # Optimize

        running_loss += loss.item()

        if (i + 1) % 100 == 0: # Log every 100 batches
            print(f'Epoch [{epoch+1}/{num_epochs}], Step [{i+1}/{len(train_loader)}], Loss: {running_loss / (i+1):.4f}')

    print(f'Epoch [{epoch+1}/{num_epochs}] finished. Average Training Loss: {running_loss / len(train_loader):.4f}')

    # Basic Validation Loop (Optional, but good practice)
    model.eval() # Set model to evaluation mode
    val_loss = 0.0
    with torch.no_grad(): # Disable gradient calculation for validation
        for inputs, labels in val_loader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            loss = criterion(outputs, labels)
            val_loss += loss.item()

    print(f'Validation Loss: {val_loss / len(val_loader):.4f}')

print("Training complete!")
````
